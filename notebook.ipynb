```python
import numpy as np
a = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([[7, 8, 9], [10, 11, 12]])
# Perform element-wise addition
result = a + b
print("Result of element-wise addition:")
print(result)   
```
